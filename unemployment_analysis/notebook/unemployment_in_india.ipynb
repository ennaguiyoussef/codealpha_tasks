```python
import pandas as pd
import numpy as np
import calendar
import plotly.graph_objects as go
import plotly.express as px

import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/raw/Unemployment_Rate_upto_11_2020.csv')
df.head()
```

### Basic Dataset Information

```python
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

print("\nColumn Names")
print(df.columns.tolist())

print("\nDataset Information")
df.info()
```

### Updating column names

```python
df.rename(columns={
    "Region": "region",
    " Date": "date",
    " Frequency": "frequency",
    " Estimated Unemployment Rate (%)": "unemployment_rate",
    " Estimated Employed": "estimated_employed",
    " Estimated Labour Participation Rate (%)": "labour_participation_rate",
    "Region.1": "zone",
    "longitude": "longitude",
    "latitude": "latitude"
}, inplace=True)

# Display updated column names
df.columns
```

- Identify duplicate records.
- Check for missing values.
- Convert the `date` column to the appropriate `datetime` data type.
- Add month and month_nb columns

```python
nb = df.duplicated().sum()
print("Number of duplicate rows in the dataset:", nb)
```

```python
mv = df.isnull().sum()
print("Number of missing values in each column:")
print(mv)
```

```python
df['date'] = pd.to_datetime(df['date'] , dayfirst=True)
df.info()
```

```python
df['month_nb'] = df['date'].dt.month
df['month'] = df['month_nb'].apply(lambda x : calendar.month_abbr[x])
df.head()
```

- Numeric data grouped by months
- Bar plot of unemployment rate and labour participation rate

```python
data = df.groupby(['month'])[['unemployment_rate' , 'estimated_employed' , 'labour_participation_rate']].mean().reset_index()
```

```python
data = (
    df.groupby(["month_nb", "month"])[
        ["unemployment_rate", "labour_participation_rate"]
    ]
    .mean()
    .reset_index()
    .sort_values("month_nb")
)

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=data["month"],
        y=data["unemployment_rate"],
        name="Unemployment Rate",
        marker_color="indianred",
        text=data["unemployment_rate"],
        texttemplate="%{text:.1f}",
        textposition="outside"
    )
)

fig.add_trace(
    go.Bar(
        x=data["month"],
        y=data["labour_participation_rate"],
        name="Labour Participation Rate",
        marker_color="lightsalmon",
        text=data["labour_participation_rate"],
        texttemplate="%{text:.1f}",
        textposition="outside"
    )
)

fig.update_layout(
    title="Average Monthly Unemployment and Labour Participation Rates",
    xaxis_title="Month",
    yaxis_title="Average Rate (%)",
    barmode="group",
    template="plotly_white"
)

fig.show()
```

- Average Estimated Employed Citizens by Month

```python
data = (
    df.groupby(["month_nb", "month"])["estimated_employed"]
      .mean()
      .reset_index()
      .sort_values("month_nb")
)

fig = px.bar(
    data_frame=data,
    x="month",
    y="estimated_employed",
    text_auto=".2s",
    color="month",
    title="Average Estimated Employed Citizens by Month",
    labels={
        "month": "Month",
        "estimated_employed": "Average Estimated Employed Citizens"
    }
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Month",
    yaxis_title="Average Estimated Employed Citizens",
    title_x=0.5
)

fig.update_traces(
    textposition="outside",
    hovertemplate="<b>%{x}</b><br>Estimated Employed: %{y:,.0f}<extra></extra>"
)

fig.show()
```

- Region-wise Unemployment Rate Analysis

```python
fig = px.box(
    df.sort_values("region"),
    x="region",
    y="unemployment_rate",
    color="region",
    points="outliers",
    title="Region-wise Distribution of Unemployment Rate"
)

fig.update_layout(
    template="plotly_white",
    xaxis_title="Region",
    yaxis_title="Unemployment Rate (%)",
)

fig.show()
```

- average unemployment rate bar plot

```python
df_region = df.groupby("region")["unemployment_rate"].mean().reset_index().sort_values('unemployment_rate', ascending=False)
fig = px.bar(
    df_region,
    x="region",
    y="unemployment_rate",
    color="region",
    title="Average Unemployment Rate by Region"
)

fig.update_layout(
    xaxis_title="Region",
    yaxis_title="Average Unemployment Rate (%)",
)

fig.show()
```

- Bar plot Unemployment Rate (monthly)

```python
fig = px.bar(
    df.sort_values("unemployment_rate", ascending=False),
    x="region",
    y="unemployment_rate",
    animation_frame="month",
    title="Average Monthly Unemployment Rate in India"
)

fig.update_layout(
    xaxis_title="Region",
    yaxis_title="Average Unemployment Rate (%)",
    title_x=0.5,
    coloraxis_showscale=False
)

fig.show()
```

```python
fig = px.scatter_geo(
    df,
    'longitude',
    'latitude',
    color='region',
    hover_name='region',
    size = 'unemployment_rate',
    scope = 'asia',
    animation_frame='month',
    title='Impact of lockdown on employment in India'
    )

fig.layout.updatemenus[0].buttons[0].args[1]['frame']['duration'] =1000
fig.update_geos(lataxis_range=[5,40],lonaxis_range=[65,100],oceancolor='lightblue',
               showocean=True)

fig.show()
```

- numeric data grouped by region

```python
data = df.groupby('zone')[['unemployment_rate', 'estimated_employed', 'labour_participation_rate']].mean().reset_index()
```

```python
fig = px.scatter_matrix(
    df,
    dimensions=['unemployment_rate', 'estimated_employed', 'labour_participation_rate'],
    color='zone',
    title='Scatter Matrix of Unemployment Rate, Estimated Employed, and Labour Participation Rate by Zone'
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5
)

fig.show()
```

```python
fig = px.bar(
    df,
    x='zone',
    y='unemployment_rate',
    color='region',
    title='Unemployment Rate by Zone and Region',
    animation_frame='month'
)

fig.update_layout(
    xaxis_title="Zone",
    yaxis_title="Average Unemployment Rate (%)",
    coloraxis_showscale=False,
    xaxis={'categoryorder':'total descending'},
    width=800,
    height=600
)
fig.layout.updatemenus[0].buttons[0].args[1]['frame']['duration'] =1000

fig.show()
```

```python
unemployment = df.groupby(['zone','region'])['unemployment_rate'].mean().reset_index()
unemployment.head()
```

```python
fig = px.sunburst(
    unemployment,
    path=['zone', 'region'],
    values='unemployment_rate',
    color='unemployment_rate',
    color_continuous_scale='RdBu',
    title='Sunburst Chart of Unemployment Rate by Zone and Region'
)

fig.show()
```
